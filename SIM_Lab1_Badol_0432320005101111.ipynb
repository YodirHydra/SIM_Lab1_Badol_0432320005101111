```python
import numpy as np
```

```python

scalar = 3.7
arr = np.array([1.2, 2.5, 3.7, 4.8, 5.6, 6.9])

print("Scalar:", scalar)
print("Array:", arr)
```

```python
print("sin:", np.sin(arr))
print("cos:", np.cos(arr))
print("tan:", np.tan(arr))

print("exp:", np.exp(arr))
print("log:", np.log(arr))
print("sqrt:", np.sqrt(arr))
print("absolute:", np.abs(arr))
```

```python
# Creating a custom 3x2 matrix
matrix_3x2 = np.array([
    [0.5, 1.2],
    [2.5, 3.7],
    [4.2, 5.5]
])

print("3x2 Matrix:")
print(matrix_3x2)
```

```python
print("Sine:")
print(np.sin(matrix_3x2))

print("\nCosine:")
print(np.cos(matrix_3x2))

print("\nTangent:")
print(np.tan(matrix_3x2))
```

```python
print("Exponential:")
print(np.exp(matrix_3x2))

print("\nNatural Logarithm:")
print(np.log(matrix_3x2))

print("\nLog base 10:")
print(np.log10(matrix_3x2))
```

```python
print("Absolute Value:")
print(np.abs(matrix_3x2))

print("\nSquare Root:")
print(np.sqrt(matrix_3x2))
```

```python
print("Remainder when divided by 2:")
print(np.mod(matrix_3x2, 2))
```

```python
print("Round:")
print(np.round(matrix_3x2))

print("\nFloor:")
print(np.floor(matrix_3x2))

print("\nCeil:")
print(np.ceil(matrix_3x2))
```

```python
# Creating a custom 2x3 matrix
matrix_2x3 = np.array([
    [2.1, 4.7, 1.5],
    [6.3, 3.2, 5.8]
])

print("2x3 Matrix:")
print(matrix_2x3)
```

```python
print("Maximum value:", np.max(matrix_2x3))
print("Minimum value:", np.min(matrix_2x3))

print("Maximum index:", np.unravel_index(np.argmax(matrix_2x3), matrix_2x3.shape))
print("Minimum index:", np.unravel_index(np.argmin(matrix_2x3), matrix_2x3.shape))
```

```python
print("Vector length:", np.linalg.norm(matrix_2x3))
```

```python
row_sorted = np.sort(matrix_2x3, axis=1)[:, ::-1]

print("Row-wise descending:")
print(row_sorted)
```

```python
entire_sorted = np.sort(matrix_2x3.flatten())[::-1]

print("Entire matrix descending:")
print(entire_sorted)
```

```python
print("Sum:", np.sum(matrix_2x3))
print("Product:", np.prod(matrix_2x3))
```

```python
print("Median:", np.median(matrix_2x3))
print("Mean:", np.mean(matrix_2x3))
print("Standard Deviation:", np.std(matrix_2x3))
```
